## 2. The question: decision, action, cost of a wrong call

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Here is a quick exploratory check of the starter dataset (content_refresh_anonymized.csv) to establish baseline numbers for our content refresh lane:

```python
import pandas as pd
try:
  df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
except FileNotFoundError:
  df = pd.read_csv('https://raw.githubusercontent.com/Zulfikar-CTRL-z/FlyRank-Internship/main/data/raw/content_refresh_anonymized.csv')
total_pages = len(df)
declining_pages = (df['trend_direction'] == 'down').sum()
pct_declining = (declining_pages / total_pages) * 100
total_impressions = df['impressions_90d'].sum()
print(f"Total Content Items: {total_pages}")
print(f"Declining Pages (trend_direction == 'down'): {declining_pages} ({pct_declining:.2f}%)")
print(f"Total 90-day Impressions: {total_impressions:,}")
```
